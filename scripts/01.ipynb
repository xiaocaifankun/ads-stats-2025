```python
print(7 / 2, 4 / 2)          # -> 3.5 2.0
print(type(7 / 2))           # -> <class 'float'>
# // 向下取整：正数好理解，但 -7 // 2 是 -4（往更小的方向取），不是 -3
print(7 // 2)                # -> 3

# % 取余：结果的符号跟随「除数」，而不是被除数
print(7 % 3) 

print( 2 * 3 )
print( 2 ** 3)
```
